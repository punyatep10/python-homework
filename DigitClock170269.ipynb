```python
from datetime import datetime

class DigitClock:
    def showTime(self):
        now = datetime.now()
        print("เวลา :", now.strftime("%H:%M:%S"))

class Calendar:
    def showDate(self):
        today = datetime.now()
        print("วันที่ :", today.strftime("%d/%m/%Y"))
```

```python
class FutureClock(DigitClock, Calendar):
    def __init__(self):
        pass

    def show(self):
        print("____________________")
        self.showDate()
        self.showTime()
        print("____________________")
```

```python
c1 = DigitClock()
c1.showTime()
```

```python
fc1 = FutureClock()
fc1.show()
```
