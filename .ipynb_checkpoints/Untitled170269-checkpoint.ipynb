```python
class Person:
    def __init__(self, cid, name, lname):
        self.cid = cid
        self.name = name
        self.lname = lname

    def info(self):
        print("{} {} {}".format(self.cid, self.name, self.lname))
```

```python
class Student(Person):
    def __init__(self, cid, name, lname):
        super().__init__(cid, name, lname)
```

```python
p1 = Person("6812732113", "Punyatep", "Areeratkun")
p1.info()
```

```python
s1 = Student("6812732113", "Punyatep", "Areesratkun")
s1.info()
```
