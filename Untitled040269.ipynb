```python
class Student:
    def __init__(self, studentID, dcn, name, last_name, age, gender):
        self.student_id = studentID
        self.name = name
        self.last_name = last_name
        self.__age = age
        self.__gender = gender
        
        if len(dcn) == 13:
            self.__dcn = dcn
        else:
            print("หมายเลขบัตรต้องมีความยาว 13 หลักเท่านั้น")
            self.__dcn = ""

    def setAge(self, newAge):
        if newAge > self.__age:
            self.__age = newAge
        else:
            print("อายุต้องเพิ่มมากกว่าเดิมเท่านั้น")

    def setDcn(self, dcn):
        if len(dcn) == 13 and self.__dcn == "":
            self.__dcn = dcn
        else:
            print("หมายเลขบัตรต้องมีความยาว 13 หลักเท่านั้น")

    def showInfo(self):
        print("""
รหัสนักศึกษา: {}
หมายเลขบัตร 13 หลัก: {}
ชื่อจริง: {}
นามสกุล: {}
อายุ: {}
เพศ: {}
        """.format(
            self.student_id,
            self.__dcn,
            self.name,
            self.last_name,
            self.__age,
            self.__gender
        ))
```

```python
s1 = Student("6812732113", "1347600002090", "ปุณยเทพ", "อารีย์รัตน์กุล", 20, "ชาย")
s1.showInfo()
s1.setAge(21)
s1.showInfo()
```
